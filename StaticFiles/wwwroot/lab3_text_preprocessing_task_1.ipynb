#### **Step 1: Preprocessing Pipeline**
Use the cell below to implement your preprocessing pipeline and create the DataFrame.



````python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from textblob import TextBlob, Word

# Download required NLTK resources (if not already done)
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab') # Added to fix LookupError

# Define the custom tweet dataset
tweets = [
    "I LUV coding tho it’s tough!! #ProgrammingLife https://t.co/abc123 🙌",
    "Rainy days r the best 4 sleeping… #cozy",
    "Sooo excited 4 the new AI class starting!!!",
    "u ever feel lyk u don’t belong??",
    "My catz r playin with toy mice 🐾 #CutePets"
]

# Your code here: Implement the preprocessing pipeline
# 1. Create a custom dictionary for standardization (at least 5 abbreviations)
# 2. Apply all preprocessing steps
# 3. Store results in a DataFrame with the specified columns

# Example structure:
# df = pd.DataFrame({
#     'original_text': tweets,
#     'preprocessed_stemmed': [...],
#     'stemmed_tokens': [...],
#     'preprocessed_lemmatized': [...],
#     'lemmatized_tokens': [...]
# })
# print(df)
abbrev_dict = {
    "luv":    "love",
    "tho":    "though",
    "r":      "are",
    "4":      "for",
    "u":      "you",
    "lyk":    "like",
    "sooo":   "so",
    "catz":   "cats",
    "playin": "playing",
    "ai":     "ai",
}

def standardize(text, dictionary):
    words = text.split()
    return " ".join(dictionary.get(w, w) for w in words)


stop_words  = set(stopwords.words('english'))
stemmer     = PorterStemmer()
lemmatizer  = WordNetLemmatizer()

def preprocess(tweet):
    """
    Returns a cleaned string ready for stemming / lemmatization.
    Steps applied:
      1. Lowercase
      2. Remove URLs
      3. Remove @mentions
      4. Remove hashtags
      5. Remove emojis / non-ASCII characters
      6. Remove punctuation & special characters
      7. Standardize abbreviations
      8. Spell-correct with TextBlob
      9. Remove stop words
    """
    text = tweet.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = text.encode('ascii', 'ignore').decode('ascii')
    text = re.sub(r"[^a-z0-9\s]", '', text)
    text = standardize(text, abbrev_dict)
    text = str(TextBlob(text).correct())
    tokens = text.split()
    tokens = [w for w in tokens if w not in stop_words]
    text = " ".join(tokens)

    return text
original_texts        = []
preprocessed_stemmed  = []
stemmed_tokens_list   = []
preprocessed_lemmed   = []
lemmed_tokens_list    = []

for tweet in tweets:
    original_texts.append(tweet)
    cleaned = preprocess(tweet)
    stemmed_words   = [stemmer.stem(w) for w in word_tokenize(cleaned)]
    stemmed_text    = " ".join(stemmed_words)
    preprocessed_stemmed.append(stemmed_text)
    stemmed_tokens_list.append(stemmed_words)
    lemmed_words  = [lemmatizer.lemmatize(w) for w in word_tokenize(cleaned)]
    lemmed_text   = " ".join(lemmed_words)
    preprocessed_lemmed.append(lemmed_text)
    lemmed_tokens_list.append(lemmed_words)

df = pd.DataFrame({
    'original_text':          original_texts,
    'preprocessed_stemmed':   preprocessed_stemmed,
    'stemmed_tokens':         stemmed_tokens_list,
    'preprocessed_lemmatized': preprocessed_lemmed,
    'lemmatized_tokens':      lemmed_tokens_list
})

pd.set_option('display.max_colwidth', None)
print(df.to_string())
````

#### **Step 2: Word Frequency Analysis**
Use the cell below to calculate and display word frequencies for both stemmed and lemmatized tokens.



````python
# Your code here: Calculate word frequencies
# 1. Extract stemmed_tokens and lemmatized_tokens from the DataFrame
# 2. Compute frequencies for both
# 3. Print the results as dictionaries or lists

# Example structure:
# from collections import Counter
# stemmed_freq = Counter([token for sublist in df['stemmed_tokens'] for token in sublist])
# lemmatized_freq = Counter([token for sublist in df['lemmatized_tokens'] for token in sublist])
# print("Stemmed Frequencies:", stemmed_freq)
# print("Lemmatized Frequencies:", lemmatized_freq)
from collections import Counter

all_stemmed_tokens    = [token for sublist in df['stemmed_tokens']    for token in sublist]
all_lemmatized_tokens = [token for sublist in df['lemmatized_tokens'] for token in sublist]

stemmed_freq    = Counter(all_stemmed_tokens)
lemmatized_freq = Counter(all_lemmatized_tokens)

print("=" * 50)
print("  Stemmed Token Frequencies")
print("=" * 50)
for word, count in stemmed_freq.most_common():
    print(f"  {word:<20} : {count}")

print()
print("=" * 50)
print("  Lemmatized Token Frequencies")
print("=" * 50)
for word, count in lemmatized_freq.most_common():
    print(f"  {word:<20} : {count}")

print()
print("Stemmed Frequencies (dict)   :", dict(stemmed_freq))
print("Lemmatized Frequencies (dict):", dict(lemmatized_freq))
````

**Stemming vs. Lemmatization — differences in word forms and frequency counts:**  
Stemming uses a rule-based algorithm (PorterStemmer) that aggressively chops word suffixes, often producing non-real-word stems such as *"excit"* (from *"excited"*), *"sleep"* (from *"sleeping"*), or *"start"* (from *"starting"*). Lemmatization, by contrast, uses a vocabulary and morphological analysis to return valid base forms — *"excited"* stays *"excited"*, *"sleeping"* becomes *"sleep"*, and *"starting"* becomes *"start"*. Because stemming collapses more variants into the same truncated token, different surface forms can merge and appear as a single, higher-count entry, whereas lemmatization maintains readable, linguistically correct forms that are easier to interpret.

**Biggest-impact preprocessing step:**  
The abbreviation standardization step (Step 4) had the greatest impact on the final output for this particular dataset. Without it, informal words like *"luv"*, *"r"*, *"4"*, *"u"*, *"lyk"*, *"sooo"*, and *"catz"* would either be spell-corrected to unintended words or remain as out-of-vocabulary tokens that carry no useful signal for a model. Replacing them with their standard equivalents (*"love"*, *"are"*, *"for"*, *"you"*, *"like"*, *"so"*, *"cats"*) ensures downstream steps such as spell correction, stop-word removal, and stemming/lemmatization all function correctly on real English words.

**Adjustments for sentiment analysis:**  
For sentiment analysis specifically, I would retain certain negation stop words — particularly *"not"*, *"no"*, *"never"*, and *"don't"* — since removing them can completely reverse the polarity of a phrase (e.g., *"don't belong"* becomes *"belong"*, losing its negative sentiment). I would also prefer lemmatization over stemming because it preserves recognizable word forms that pre-trained sentiment lexicons (e.g., VADER, SentiWordNet) are built on; stems like *"excit"* may not match lexicon entries. Finally, I would expand the custom dictionary to capture more sentiment-bearing slang (e.g., *"gr8"* → *"great"*, *"omg"* → *"oh my god"*) to ensure emotionally charged informal language is not discarded.
